```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import pandas as pd
import time

url = "https://coinmarketcap.com/"

driver = webdriver.Chrome()
driver.get(url)

time.sleep(10)

body = driver.find_element(By.CLASS_NAME, "cmc-body-wrapper")

html = body.get_attribute("innerHTML")

soup = BeautifulSoup(html, "html.parser")

table = soup.find("table")

if table:
    df = pd.read_html(str(table))[0]
    display(df)
    df.to_csv("coinmarketcap.csv", index=False)
    print("Data saved successfully!")
else:
    print("Table not found!")

driver.quit()
```

```python

import pyodbc

conn = pyodbc.connect(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=HAMMAD-AFRIDI;"
    "DATABASE=CryptoDB;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

cursor = conn.cursor()

try:
    cursor.execute("EXEC Load_Crypto_History")
    conn.commit()

    cursor.execute("""
        INSERT INTO ETL_Log
        (Pipeline_Name, Status, Message)
        VALUES (?, ?, ?)
    """, ("CoinMarketCap_ETL", "SUCCESS", "Data loaded successfully"))

    conn.commit()
    print("ETL Pipeline Successful!")

except Exception as e:
    conn.rollback()

    cursor.execute("""
        INSERT INTO ETL_Log
        (Pipeline_Name, Status, Message)
        VALUES (?, ?, ?)
    """, ("CoinMarketCap_ETL", "FAILED", str(e)[:500]))

    conn.commit()
    print("ETL Failed:", e)

finally:
    cursor.close()
    conn.close()
```

```python
import pyodbc
import logging

logging.basicConfig(
    filename="etl_error.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

try:
    conn = pyodbc.connect(
        "DRIVER={ODBC Driver 18 for SQL Server};"
        "SERVER=HAMMAD-AFRIDI;"
        "DATABASE=CryptoDB;"
        "Trusted_Connection=yes;"
        "TrustServerCertificate=yes;"
    )

    cursor = conn.cursor()

    cursor.execute("EXEC Load_Crypto_History")
    conn.commit()

    logging.info("ETL completed successfully")
    print("ETL SUCCESS")

except Exception as e:
    logging.exception("ETL FAILED")
    print("ETL FAILED:", e)

finally:
    if 'conn' in locals():
        conn.close()
```
